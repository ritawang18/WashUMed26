```python
# Parse all Batch Pre-Scan TXT reports and display INSIDE ROI + WHOLE TISSUE statistics
# Requires pandas when running the notebook. Falls back to a brief message if pandas isn't available.
from pathlib import Path
import re

# Adjust these paths if your environment differs
DOWNLOADS_ROOT = Path(r"C:\Users\hanss\Downloads\DEXA Scans (1)\DEXA Scans")
BATCH_PREFIX = 'Batch '

# Patterns to extract numbers
pat_ref = re.compile(r"Reference Image:\s*(.+)")
pat = lambda name: re.compile(rf"{re.escape(name)}:\s*([0-9.]+)")
patterns = {
    'sample_area': re.compile(r"Sample Area:\s*([0-9.]+)\s*cm\^2"),
    'bone_area': re.compile(r"Bone Area:\s*([0-9.]+)\s*cm\^2"),
    'total_weight': re.compile(r"Total Weight:\s*([0-9.]+)\s*g"),
    'soft_weight': re.compile(r"Soft Weight:\s*([0-9.]+)\s*g"),
    'lean_weight': re.compile(r"Lean Weight:\s*([0-9.]+)\s*g"),
    'fat_weight': re.compile(r"Fat Weight:\s*([0-9.]+)\s*g"),
    'fat_percent': re.compile(r"Fat Percent:\s*([0-9.]+)"),
    'BMC': re.compile(r"BMC:\s*([0-9.]+)\s*g"),
    'BMD': re.compile(r"BMD:\s*([0-9.]+)\s*mg/cm\^2"),
}

rows = []
for batch_dir in sorted(DOWNLOADS_ROOT.glob('Batch *')):
    pre = batch_dir / 'Pre-Scan'
    if not pre.exists():
        continue
    for sex in ['Male', 'Female']:
        folder = pre / sex
        if not folder.exists():
            continue
        for txt in sorted(folder.glob('*.txt')):
            text = txt.read_text(encoding='utf-8', errors='replace')
            row = {
                'batch': batch_dir.name,
                'sex': sex,
                'filename': txt.name,
                'reference_image': '',
            }
            m = pat_ref.search(text)
            if m:
                row['reference_image'] = m.group(1).strip()

            # split INSIDE and WHOLE sections
            inside_block = ''
            whole_block = ''
            if 'INSIDE ROI TISSUE STATISTICS:' in text and 'WHOLE TISSUE STATISTICS:' in text:
                inside_block = text.split('INSIDE ROI TISSUE STATISTICS:')[1].split('WHOLE TISSUE STATISTICS:')[0]
                whole_block = text.split('WHOLE TISSUE STATISTICS:')[1]
            else:
                # fallback: attempt to parse whole file as inside
                inside_block = text

            # extract values
            for k, p in patterns.items():
                mi = p.search(inside_block)
                mw = p.search(whole_block)
                row[f'inside_{k}'] = mi.group(1) if mi else ''
                row[f'whole_{k}'] = mw.group(1) if mw else ''
            rows.append(row)

# Display with pandas if available
try:
    import pandas as pd
    df = pd.DataFrame(rows)
    # Reorder columns for readability
    cols = ['batch','sex','filename','reference_image'] + \
           [c for c in df.columns if c.startswith('inside_')] + \
           [c for c in df.columns if c.startswith('whole_')]
    cols = [c for c in cols if c in df.columns]
    display(df[cols].sort_values(['batch','sex','filename']).reset_index(drop=True))
except Exception as e:
    print('pandas is not available in this kernel. Parsed rows count:', len(rows))
    # Print header and first 5 rows
    if rows:
        keys = list(rows[0].keys())
        print(','.join(keys))
        for r in rows[:5]:
            print(','.join(str(r.get(k,'')) for k in keys))
    print('\nTo get a nice table, run this notebook with a Python kernel that has pandas installed. Error:', e)
```
