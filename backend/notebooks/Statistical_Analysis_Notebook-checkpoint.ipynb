```python
import pandas as pd
import re
from scipy import stats
import numpy as np

# Dataset
df = pd.read_csv(
    "data/Review_SY-08002944_4_3_2025 10_31_21_cleaned.csv",
    sep=",",
    quotechar='"',
    engine="python",
    on_bad_lines="skip"
)
```

```python
# --- Clean Numeric Values ---
def clean_numeric(x):
    """Remove text flags (e.g. 'H', 'L', '*') and keep only numeric values."""
    if isinstance(x, str):
        x = re.sub(r"[^0-9.\-]", "", x)
        return float(x) if x else np.nan
    return x

numeric = [
    c for c in df.columns
    if any(sym in c for sym in ["#", "%", "(", ")", "/", "HGB", "RBC", "WBC", "HCT", "MCH", "MCV", "PLT"])
]

for col in numeric:
    df[col] = df[col].apply(clean_numeric)

#---Dependent Variable
df["Neutrophils"] = df["Neu # (10^3/uL)"]
```

```python
#---Potential Candidates---

for col in ["WBC Message", "RBC Message", "PLT Message"]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace(r"[\n\r]", " ", regex=True).str.strip()
        df[col].replace(["nan", ""], np.nan, inplace=True)

#---Biologically meaningful elements---
possible_factors = ["Operator", "WBC Message", "RBC Message"]
relevant = []

for col in possible_factors:
    subset = df.dropna(subset=[col, "Neutrophils"])
    unique_vals = subset[col].nunique()
    if unique_vals >= 2:
        relevant.append(col)

print("Eligible columns for ANOVA:", relevant)
```

```python
# ---ANOVA RESULTS---
anova_results = {}

for col in relevant:
    subset = df.dropna(subset=[col, "Neutrophils"])
    groups = [vals["Neutrophils"].values for _, vals in subset.groupby(col)]
    if len(groups) >= 2:
        # Perform one-way ANOVA
        f_stat, p_value = stats.f_oneway(*groups)
        
        # Compute degrees of freedom
        df_between = len(groups) - 1
        df_within = len(subset) - len(groups)
        df_total = len(subset) - 1
        
        anova_results[col] = {
            "F-statistic": f_stat,
            "p-value": p_value,
            "groups": len(groups),
            "df_between": df_between,
            "df_within": df_within,
            "df_total": df_total
        }

print("\n--- ANOVA Results ---")
if anova_results:
    for k, v in anova_results.items():
        print(
            f"{k}: F={v['F-statistic']:.4f}, p={v['p-value']:.6f}, "
            f"groups={v['groups']}, df_between={v['df_between']}, df_within={v['df_within']}, df_total={v['df_total']}"
        )
else:
    print("No categorical variables met ANOVA criteria.")
```

```python
#---Identify Numeric Columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()


# Drop columns with all NaN or no variance
valid = [c for c in numeric_cols if df[c].notna().sum() > 2 and df[c].nunique() > 1]
print("\nNumeric columns used for correlation:", valid)
```

```python
# --- PEARSON ---
pearson_results = []
for i in range(len(valid)):
    for j in range(i + 1, len(valid)):
        col1, col2 = valid[i], valid[j]
        subset = df[[col1, col2]].dropna()
        if len(subset) > 2:
            try:
                r_val, p_val = stats.pearsonr(subset[col1], subset[col2])
                pearson_results.append({
                    "Variable 1": col1,
                    "Variable 2": col2,
                    "r": r_val,
                    "p-value": p_val,
                    "n": len(subset)
                })
            except Exception:
                continue

pearson_df = pd.DataFrame(pearson_results)
if not pearson_df.empty:
    print("\n--- Top 10 Pearson Correlations ---")
    print(pearson_df.sort_values(by="r", ascending=False).head(10)[["Variable 1", "Variable 2", "r", "p-value", "n"]])
else:
    print("\nNo valid numeric pairs for Pearson correlation.")
```
